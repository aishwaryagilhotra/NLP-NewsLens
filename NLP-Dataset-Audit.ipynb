```python

from collections import Counter

entity_counts = Counter()

for split in ["train", "validation", "test"]:
    counts = Counter()

    for example in dataset[split]:
        for tag in example["tags"]:
            label = id2label[tag]

            if label.startswith("B-"):
                counts[label[2:]] += 1

    entity_counts[split] = counts

for split, counts in entity_counts.items():
    print(f"\n{split.upper()}")
    for entity_type, count in sorted(counts.items()):
        print(f"{entity_type:15s} {count:>7,}")
```

```python

from collections import Counter

def validate_bio_split(split_name, max_examples=5):
    invalid = []
    total_tokens = 0

    for row_idx, example in enumerate(dataset[split_name]):
        tokens = example["tokens"]
        tags = example["tags"]
        previous_type = None

        total_tokens += len(tokens)

        for pos, tag_id in enumerate(tags):
            label = id2label[tag_id]

            if label == "O":
                previous_type = None
                continue

            prefix, entity_type = label.split("-", 1)

            if prefix == "B":
                previous_type = entity_type

            elif prefix == "I":
                if previous_type != entity_type:
                    invalid.append({
                        "row": row_idx,
                        "position": pos,
                        "token": tokens[pos],
                        "label": label,
                        "previous_type": previous_type,
                    })

                previous_type = entity_type

    print(f"\n{split_name.upper()}")
    print("Tokens:", f"{total_tokens:,}")
    print("Invalid I-tag transitions:", len(invalid))

    for issue in invalid[:max_examples]:
        print(issue)

    return invalid


invalid_train = validate_bio_split("train")
invalid_val = validate_bio_split("validation")
invalid_test = validate_bio_split("test")
```

```python

def inspect_anomalies(split_name, anomalies, limit=10, window=5):
    split = dataset[split_name]

    for issue in anomalies[:limit]:
        row_idx = issue["row"]
        pos = issue["position"]
        example = split[row_idx]

        start = max(0, pos - window)
        end = min(len(example["tokens"]), pos + window + 1)

        print(
            f"\nSplit: {split_name} | Row: {row_idx} "
            f"| Position: {pos}"
        )

        for i in range(start, end):
            token = example["tokens"][i]
            label = id2label[example["tags"][i]]
            marker = " <-- ANOMALY" if i == pos else ""
            print(f"{i:>3}  {token:<20} {label}{marker}")


inspect_anomalies("train", invalid_train, limit=10)
```

```python

def extract_entities(tokens, tags, id2label):
    """
    Reconstruct entity spans from token-level BIO annotations.

    Malformed I-TYPE tags start a new span of TYPE.
    Original tokens and tags remain unchanged.
    End indices are exclusive.
    """
    entities = []
    start = None
    current_type = None

    def close_entity(end):
        if start is not None:
            entities.append({
                "text": " ".join(tokens[start:end]),
                "label": current_type,
                "start_token": start,
                "end_token": end,
            })

    for i, tag_id in enumerate(tags):
        label = id2label[tag_id]

        if label == "O":
            close_entity(i)
            start = None
            current_type = None
            continue

        prefix, entity_type = label.split("-", 1)

        if prefix == "B":
            close_entity(i)
            start = i
            current_type = entity_type

        elif prefix == "I":
            if current_type != entity_type:
                # Recover from an orphan or mismatched I-tag.
                close_entity(i)
                start = i
                current_type = entity_type

    close_entity(len(tokens))
    return entities


# Test on one anomalous example
example = dataset["train"][138]

for entity in extract_entities(
    example["tokens"],
    example["tags"],
    id2label
):
    print(entity)
```

```python

from collections import Counter

def analyze_split(split_name):
    entity_counts = Counter()
    total_entities = 0
    total_sequences = len(dataset[split_name])

    for example in dataset[split_name]:
        entities = extract_entities(
            example["tokens"],
            example["tags"],
            id2label
        )

        total_entities += len(entities)

        for entity in entities:
            entity_counts[entity["label"]] += 1

    print(f"\n{split_name.upper()}")
    print("Sequences:", f"{total_sequences:,}")
    print("Reconstructed entities:", f"{total_entities:,}")

    for label, count in entity_counts.most_common():
        print(f"{label:15s} {count:>8,}")

    return entity_counts


train_entity_counts = analyze_split("train")
val_entity_counts = analyze_split("validation")
test_entity_counts = analyze_split("test")
```
